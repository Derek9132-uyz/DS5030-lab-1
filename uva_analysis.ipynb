```python
import pandas as pd
import numpy as np
import seaborn as sb
import matplotlib.pyplot as plt
import os
```

```python
foreign_gifts = pd.read_csv("ForeignGifts_edu.csv")
uva_data = foreign_gifts.loc[foreign_gifts["Institution Name"] == "University of Virginia"]
uva_data.reset_index(inplace=True)
uva_data.head()
```

**ANALYSIS METHODS:** <br>
1 ) Find total amount given to UVA <br>
2 ) Find biggest gifter to UVA <br>
3 ) Find value counts of gift types <br>
4 ) Find mean, median, variance <br>
5 ) Compare with totals, means, medians of 2 other Virginia schools (William & Mary, Virginia Tech) and 3 outside schools <br>


```python
wm_data = foreign_gifts.loc[foreign_gifts["Institution Name"] == "College of William & Mary"]
vt_data = foreign_gifts.loc[foreign_gifts["Institution Name"] == "Virginia Polytechnic Institute & State University"]

upen_data = foreign_gifts.loc[foreign_gifts["Institution Name"] == "University of Pennsylvania"]
unc_data = foreign_gifts.loc[foreign_gifts["Institution Name"] == "University of North Carolina - Chapel Hill"]
umd_data = foreign_gifts.loc[foreign_gifts["Institution Name"] == "University of Maryland, College Park"]
```

```python
len(uva_data)
```

```python
uva_sum = uva_data["Foreign Gift Amount"].sum()
uva_sum
```

```python
uva_mean = uva_data["Foreign Gift Amount"].mean()
uva_mean
```

```python
uva_med = uva_data["Foreign Gift Amount"].median()
uva_med
```

```python
uva_data.groupby(["Country of Giftor"])["Foreign Gift Amount"].sum().sort_values(ascending=False)
```

Bangladesh is the biggest giftor to UVA.

```python
uva_data["Gift Type"].value_counts()
```

UVA has no real estate gifts.

```python
wm_num_gifts = len(wm_data)
wm_sum = wm_data["Foreign Gift Amount"].sum()
wm_mean = wm_data["Foreign Gift Amount"].mean()
wm_med = wm_data["Foreign Gift Amount"].median()

print("William & Mary received " + str(wm_num_gifts) + " gifts.")
print("William & Mary received $" + str(wm_sum) + " in total.")
print("William & Mary's gift mean was " + str(wm_mean))
print("William & Mary's gift median was " + str(wm_med))
```
