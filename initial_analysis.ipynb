```python
foreign_gifts["Gift Type"].value_counts(normalize=True)
```

2 ) <br>
The three gift types are contract, monetary gift, and real estate. A gift in a contract is part of some sort of agreement, perhaps an agreement to a mutual exchange of gifts or something similar. A monetary gift is a simple gift of money from one party to another with (presumably) no strings attached. A gift in the form of real estate is the gifting of a property or building from one party to another.

```python
foreign_gifts.groupby(["Country of Giftor"])["Foreign Gift Amount"].sum().sort_values(ascending=False)
```

3-1 ) <br>
Qatar gives the most money in total.

```python
foreign_gifts.groupby(["Institution Name"])["Foreign Gift Amount"].median().sort_values(ascending=False)
```

3-6 ) <br>
In the displayed data, the mean is either close to or much larger than the median. In situations where the mean is much larger, it is likely due to outlier values that are much higher than the rest of the values, which skew the data towards them.

```python
foreign_gifts.groupby(["Country of Giftor", "Institution Name"])["Foreign Gift Amount"].sum().sort_values(ascending=False)
```
